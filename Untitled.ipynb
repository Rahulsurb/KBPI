```python
################Step 1: Reproject All Soil Rasters


import os
import rasterio
from rasterio.warp import calculate_default_transform, reproject, Resampling

# ---------------------------------------------------
# Input & Output folders
# ---------------------------------------------------

input_folder = r"D:\Khet Bachao\soil data"
output_folder = r"D:\Khet Bachao\soil data\Projected"

os.makedirs(output_folder, exist_ok=True)

target_crs = "EPSG:4326"

# ---------------------------------------------------
# Process all TIFFs
# ---------------------------------------------------

for file in os.listdir(input_folder):

    if not file.lower().endswith(".tif"):
        continue

    input_raster = os.path.join(input_folder, file)
    output_raster = os.path.join(output_folder, file)

    with rasterio.open(input_raster) as src:

        print(f"\nProcessing : {file}")
        print("Current CRS :", src.crs)

        # Already in EPSG:4326
        if str(src.crs) == target_crs:

            print("Already EPSG:4326 → Copying...")

            profile = src.profile

            with rasterio.open(output_raster, "w", **profile) as dst:

                for i in range(1, src.count + 1):
                    dst.write(src.read(i), i)

            continue

        transform, width, height = calculate_default_transform(
            src.crs,
            target_crs,
            src.width,
            src.height,
            *src.bounds
        )

        kwargs = src.meta.copy()

        kwargs.update({
            "crs": target_crs,
            "transform": transform,
            "width": width,
            "height": height
        })

        with rasterio.open(output_raster, "w", **kwargs) as dst:

            for i in range(1, src.count + 1):

                reproject(
                    source=rasterio.band(src, i),
                    destination=rasterio.band(dst, i),
                    src_transform=src.transform,
                    src_crs=src.crs,
                    dst_transform=transform,
                    dst_crs=target_crs,
                    resampling=Resampling.bilinear
                )

print("\nAll rasters are ready.")
```

```python
import os
import geopandas as gpd
import rasterio
from rasterstats import zonal_stats
import pandas as pd

# ---------------------------------------------------------
# Paths
# ---------------------------------------------------------

shapefile = r"D:\Khet Bachao\BU_District\Export_Output.shp"

soil_folder = r"D:\Khet Bachao\soil data"

output_csv = r"D:\Khet Bachao\soil data\Bundlekhand_Soil_Properties.csv"

# ---------------------------------------------------------
# Read & Reproject
# ---------------------------------------------------------

gdf = gpd.read_file(shapefile)
gdf = gdf.to_crs("EPSG:4326")

# ---------------------------------------------------------
# Soil rasters
# ---------------------------------------------------------

soil_files = {
    "Bulk_Density": "BDOD_0_30cm.tif",
    "CEC": "CEC_0_30cm.tif",
    "Clay": "CLAY_0_30cm.tif",
    "Nitrogen": "NITROGEN_0_30cm.tif",
    "pH": "PH_0_30cm.tif",
    "SOC": "SOC_0_30cm.tif"
}

# ---------------------------------------------------------
# Extract zonal statistics
# ---------------------------------------------------------

for variable, raster_name in soil_files.items():

    raster = os.path.join(soil_folder, raster_name)

    print(f"Processing {variable}")

    stats = zonal_stats(
        vectors=gdf,
        raster=raster,
        stats=["mean"],
        all_touched=True
    )

    gdf[variable] = [x["mean"] for x in stats]

# ---------------------------------------------------------
# Final table
# ---------------------------------------------------------

final = gdf[
    [
        "STATE_UT",
        "DISTRICT",
        "Bulk_Density",
        "CEC",
        "Clay",
        "Nitrogen",
        "pH",
        "SOC"
    ]
].copy()

final.rename(columns={
    "STATE_UT":"State",
    "DISTRICT":"District"
}, inplace=True)

# Standardize names
final["State"] = final["State"].str.title().str.strip()
final["District"] = final["District"].str.title().str.strip()

# Round values
soil_cols = [
    "Bulk_Density",
    "CEC",
    "Clay",
    "Nitrogen",
    "pH",
    "SOC"
]

final[soil_cols] = final[soil_cols].round(3)

print(final)

final.to_csv(output_csv,index=False)

print("\nSaved:",output_csv)
```

```python
################ phase 4

# ==========================================================
# PHASE IV
# Data Normalization for Entropy Weight Method
# ==========================================================

import pandas as pd
import numpy as np

# ==========================================================
# Read Dataset
# ==========================================================

df = pd.read_excel("D:\Khet Bachao\DATA.xlsx")

# ==========================================================
# Final Indicators
# ==========================================================

data = df[[
    "State",
    "District",
    "Mean_Rainfall_mm",
    "Mean_Soil_Moisture",
    "Mean_NDVI",
    "Mean_NDMI",
    "Bulk_Density",
    "CEC",
    "Nitrogen",
    "SOC",
    "Irrigation_Coverage",
    "Cropping_Intensity"
]]

# ==========================================================
# Indicator Types
# ==========================================================

benefit = [

    "Mean_Rainfall_mm",
    "Mean_Soil_Moisture",
    "Mean_NDVI",
    "Mean_NDMI",
    "CEC",
    "Nitrogen",
    "SOC",
    "Irrigation_Coverage",
    "Cropping_Intensity"

]

cost = [

    "Bulk_Density"

]

# ==========================================================
# Create Normalized Dataset
# ==========================================================

normalized = data.copy()

# ==========================================================
# Benefit Indicators
# ==========================================================

for col in benefit:

    normalized[col] = (

        data[col] - data[col].min()

    ) / (

        data[col].max() - data[col].min()

    )

# ==========================================================
# Cost Indicators
# ==========================================================

for col in cost:

    normalized[col] = (

        data[col].max() - data[col]

    ) / (

        data[col].max() - data[col].min()

    )

# ==========================================================
# Display Results
# ==========================================================

print("\nNormalized Dataset")

print(normalized.round(4))

# ==========================================================
# Save
# ==========================================================

normalized.to_excel(

    "PhaseIV_Normalized_Data.xlsx",

    index=False

)

print("\nNormalization Completed Successfully.")
```

```python
############### PHASE 5

# ==========================================================
# PHASE V
# Entropy Weight Method (EWM)
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"

# ==========================================================
# Read Normalized Dataset
# ==========================================================

df = pd.read_excel("PhaseIV_Normalized_Data.xlsx")

# ==========================================================
# Indicators
# ==========================================================

indicator_cols = [

    "Mean_Rainfall_mm",
    "Mean_Soil_Moisture",
    "Mean_NDVI",
    "Mean_NDMI",
    "Bulk_Density",
    "CEC",
    "Nitrogen",
    "SOC",
    "Irrigation_Coverage",
    "Cropping_Intensity"

]

X = df[indicator_cols].copy()

# ==========================================================
# STEP 1
# Probability Matrix
# ==========================================================

P = X.div(X.sum(axis=0), axis=1)

# Avoid log(0)
P = P.replace(0, 1e-12)

P.to_excel(
    "PhaseV_Probability_Matrix.xlsx",
    index=False
)

print("\nProbability Matrix")
print(P.round(5))

# ==========================================================
# STEP 2
# Entropy
# ==========================================================

n = len(P)

k = 1 / np.log(n)

entropy = -k * (P * np.log(P)).sum(axis=0)

# ==========================================================
# STEP 3
# Diversification
# ==========================================================

diversification = 1 - entropy

# ==========================================================
# STEP 4
# Entropy Weights
# ==========================================================

weights = diversification / diversification.sum()

# ==========================================================
# Final Table
# ==========================================================

ewm = pd.DataFrame({

    "Indicator": indicator_cols,

    "Entropy": entropy.values,

    "Diversification": diversification.values,

    "Weight": weights.values

})

ewm = ewm.sort_values(
    by="Weight",
    ascending=False
)

print("\nEntropy Weight Results\n")
print(ewm.round(4))

ewm.to_excel(
    "PhaseV_Entropy_Weights.xlsx",
    index=False
)
```

```python
# ==========================================================
# PHASE VI
# Agricultural Resource Index (ARI)
# & Khet Bachao Priority Index (KBPI)
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"

# ==========================================================
# Read Normalized Dataset
# ==========================================================

df = pd.read_excel("PhaseIV_Normalized_Data.xlsx")

# ==========================================================
# Entropy Weights
# ==========================================================

weights = {

    "Mean_Rainfall_mm": 0.0887,
    "Mean_Soil_Moisture": 0.1952,
    "Mean_NDVI": 0.0670,
    "Mean_NDMI": 0.0923,
    "Bulk_Density": 0.0816,
    "CEC": 0.2266,
    "Nitrogen": 0.1103,
    "SOC": 0.0570,
    "Irrigation_Coverage": 0.0387,
    "Cropping_Intensity": 0.0425

}

# ==========================================================
# STEP 1
# Agricultural Resource Index (ARI)
# ==========================================================

df["ARI"] = 0.0

for indicator, weight in weights.items():
    df["ARI"] += df[indicator] * weight

# ==========================================================
# STEP 2
# Khet Bachao Priority Index (KBPI)
# ==========================================================

df["KBPI"] = 1 - df["ARI"]

# ==========================================================
# STEP 3
# Ranking
# ==========================================================

df["ARI_Rank"] = df["ARI"].rank(
    ascending=False,
    method="dense"
).astype(int)

df["KBPI_Rank"] = df["KBPI"].rank(
    ascending=False,
    method="dense"
).astype(int)

# ==========================================================
# Sort by Priority
# ==========================================================

priority_df = df.sort_values(
    by="KBPI",
    ascending=False
)

# ==========================================================
# Save
# ==========================================================

priority_df.to_excel(
    "PhaseVI_ARI_KBPI.xlsx",
    index=False
)

print(priority_df[
    [
        "State",
        "District",
        "ARI",
        "ARI_Rank",
        "KBPI",
        "KBPI_Rank"
    ]
].round(4))
```

```python
# ==========================================================
# Phase VIII
# Sensitivity Analysis
# Entropy Weight vs Equal Weight
# ==========================================================

import pandas as pd
import numpy as np

from scipy.stats import spearmanr

import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"

# ==========================================================
# Read Files
# ==========================================================

# Normalized indicators
norm = pd.read_excel("PhaseIV_Normalized_Data.xlsx")

# Final entropy results
entropy = pd.read_excel("PhaseVI_ARI_KBPI.xlsx")

# ==========================================================
# Indicator List
# ==========================================================

indicators = [

    "Mean_Rainfall_mm",

    "Mean_Soil_Moisture",

    "Mean_NDVI",

    "Mean_NDMI",

    "Bulk_Density",

    "CEC",

    "Nitrogen",

    "SOC",

    "Irrigation_Coverage",

    "Cropping_Intensity"

]

# ==========================================================
# Equal Weight Method
# ==========================================================

equal_weight = 1 / len(indicators)

print("Equal Weight =", equal_weight)

# Agricultural Resource Index

norm["ARI_Equal"] = norm[indicators].mean(axis=1)

# KBPI

norm["KBPI_Equal"] = 1 - norm["ARI_Equal"]

# ==========================================================
# Merge with Entropy Results
# ==========================================================

df = entropy.merge(

    norm[["District","ARI_Equal","KBPI_Equal"]],

    on="District",

    how="left"

)

# ==========================================================
# Equal Weight Ranking
# ==========================================================

df["Equal_Rank"] = (

    df["KBPI_Equal"]

      .rank(

          ascending=False,

          method="dense"

      )

      .astype(int)

)

# ==========================================================
# Rank Difference
# ==========================================================

df["Rank_Change"] = (

    df["KBPI_Rank"] -

    df["Equal_Rank"]

)

df["Absolute_Change"] = (

    df["Rank_Change"]

    .abs()

)

# ==========================================================
# Spearman Rank Correlation
# ==========================================================

rho, pvalue = spearmanr(

    df["KBPI_Rank"],

    df["Equal_Rank"]

)

print()

print("====================================")

print("Spearman Rank Correlation")

print("====================================")

print("rho =", round(rho,4))

print("p-value =", round(pvalue,6))

# ==========================================================
# Mean Absolute Rank Difference
# ==========================================================

mard = df["Absolute_Change"].mean()

print()

print("Mean Absolute Rank Difference =", round(mard,3))

# ==========================================================
# Sort Table
# ==========================================================

comparison = df[[

    "State",

    "District",

    "KBPI",

    "KBPI_Rank",

    "KBPI_Equal",

    "Equal_Rank",

    "Rank_Change",

    "Absolute_Change"

]]

comparison = comparison.sort_values(

    "KBPI_Rank"

)

print()

print(comparison)

# ==========================================================
# Export Excel
# ==========================================================

comparison.to_excel(

    "Sensitivity_Analysis.xlsx",

    index=False

)

print()

print("Sensitivity_Analysis.xlsx exported.")

# ==========================================================
# Publication Figure
# Ranking Comparison
# ==========================================================

comparison_plot = comparison.sort_values(

    "KBPI_Rank"

)

x = np.arange(len(comparison_plot))

width = 0.35

fig, ax = plt.subplots(

    figsize=(10,6)

)

ax.bar(

    x-width/2,

    comparison_plot["KBPI_Rank"],

    width,

    label="Entropy Weight"

)

ax.bar(

    x+width/2,

    comparison_plot["Equal_Rank"],

    width,

    label="Equal Weight"

)

ax.set_xticks(x)

ax.set_xticklabels(

    comparison_plot["District"],

    rotation=45,

    ha="right"

)

ax.set_ylabel(

    "District Rank",

    fontsize=13,

    fontweight="bold"

)

ax.set_xlabel(

    "District",

    fontsize=13,

    fontweight="bold"

)

ax.set_title(

    "Sensitivity Analysis of District Rankings",

    fontsize=15,

    fontweight="bold"

)

ax.legend()

plt.tight_layout()

plt.savefig(

    "Sensitivity_Ranking_Comparison.tif",

    dpi=600,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Summary
# ==========================================================

print()

print("====================================")

print("Sensitivity Analysis Summary")

print("====================================")

print("Spearman rho :", round(rho,4))

print("Mean Absolute Rank Difference :", round(mard,3))

print("Maximum Rank Change :", df["Absolute_Change"].max())

print("Minimum Rank Change :", df["Absolute_Change"].min())
```

```python
################ Complete Python Script: Leave-One-Indicator-Out (LOO) Sensitivity Analysis

# ==========================================================
# Leave-One-Indicator-Out (LOO) Sensitivity Analysis
# Khet Bachao Priority Index (KBPI)
# ==========================================================

import pandas as pd
import numpy as np
from scipy.stats import spearmanr
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"

# ==========================================================
# Read normalized dataset
# ==========================================================

df = pd.read_excel("PhaseIV_Normalized_Data.xlsx")

# Original KBPI Results
original = pd.read_excel("PhaseVI_ARI_KBPI.xlsx")

# ==========================================================
# Indicator List
# ==========================================================

indicators = [

    "Mean_Rainfall_mm",
    "Mean_Soil_Moisture",
    "Mean_NDVI",
    "Mean_NDMI",
    "Bulk_Density",
    "CEC",
    "Nitrogen",
    "SOC",
    "Irrigation_Coverage",
    "Cropping_Intensity"

]

# ==========================================================
# Entropy Weight Function
# ==========================================================

def entropy_weights(data):

    X = data.copy()

    # Avoid division by zero
    X = X.replace(0, 1e-12)

    # Probability matrix
    P = X.div(X.sum(axis=0), axis=1)

    n = len(X)

    k = 1 / np.log(n)

    entropy = -k * (P * np.log(P)).sum(axis=0)

    diversification = 1 - entropy

    weights = diversification / diversification.sum()

    return weights

# ==========================================================
# Store Results
# ==========================================================

summary = []

all_rankings = []

# ==========================================================
# Loop through indicators
# ==========================================================

for removed in indicators:

    print("-------------------------------------------")
    print("Removing:", removed)

    remaining = [x for x in indicators if x != removed]

    temp = df[remaining].copy()

    # --------------------------------------
    # Recalculate entropy weights
    # --------------------------------------

    weights = entropy_weights(temp)

    # --------------------------------------
    # Agricultural Resource Index
    # --------------------------------------

    ari = (temp * weights).sum(axis=1)

    kbpi = 1 - ari

    rank = (
        kbpi.rank(
            ascending=False,
            method="dense"
        )
        .astype(int)
    )

    result = pd.DataFrame({

        "District": df["District"],
        "Removed": removed,
        "KBPI": kbpi,
        "Rank": rank

    })

    all_rankings.append(result)

    # --------------------------------------
    # Merge with original
    # --------------------------------------

    compare = original.merge(

        result,

        on="District"

    )

    compare["Rank_Diff"] = (

        compare["KBPI_Rank"]

        -

        compare["Rank"]

    )

    compare["Abs_Diff"] = (

        compare["Rank_Diff"]

        .abs()

    )

    # --------------------------------------
    # Statistics
    # --------------------------------------

    rho, p = spearmanr(

        compare["KBPI_Rank"],

        compare["Rank"]

    )

    mean_change = compare["Abs_Diff"].mean()

    max_change = compare["Abs_Diff"].max()

    sensitive = (compare["Abs_Diff"] >= 2).sum()

    summary.append({

        "Removed_Indicator": removed,

        "Spearman_rho": round(rho,4),

        "P_value": p,

        "Mean_Absolute_Rank_Difference": round(mean_change,3),

        "Maximum_Rank_Change": int(max_change),

        "Districts_with_RankChange>=2": int(sensitive)

    })

# ==========================================================
# Summary Table
# ==========================================================

summary_df = pd.DataFrame(summary)

summary_df = summary_df.sort_values(

    "Spearman_rho",

    ascending=False

)

print(summary_df)

# ==========================================================
# Export
# ==========================================================

summary_df.to_excel(

    "LOO_Sensitivity_Summary.xlsx",

    index=False

)

rankings = pd.concat(all_rankings)

rankings.to_excel(

    "LOO_All_Rankings.xlsx",

    index=False

)

print()

print("Files Exported")

print("1. LOO_Sensitivity_Summary.xlsx")

print("2. LOO_All_Rankings.xlsx")

# ==========================================================
# Publication Figure
# ==========================================================

plt.figure(figsize=(9,5))

plt.barh(

    summary_df["Removed_Indicator"],

    summary_df["Spearman_rho"]

)

plt.xlabel(

    "Spearman Rank Correlation",

    fontsize=12,

    fontweight="bold"

)

plt.ylabel(

    "Removed Indicator",

    fontsize=12,

    fontweight="bold"

)

plt.title(

    "Leave-One-Indicator-Out Sensitivity Analysis",

    fontsize=14,

    fontweight="bold"

)

plt.xlim(0.7,1.0)

plt.tight_layout()

plt.savefig(

    "LOO_Spearman.tif",

    dpi=600,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Additional Figure
# ==========================================================

plt.figure(figsize=(9,5))

plt.barh(

    summary_df["Removed_Indicator"],

    summary_df["Mean_Absolute_Rank_Difference"]

)

plt.xlabel(

    "Mean Absolute Rank Difference",

    fontsize=12,

    fontweight="bold"

)

plt.ylabel(

    "Removed Indicator",

    fontsize=12,

    fontweight="bold"

)

plt.title(

    "Average Rank Change after Removing Each Indicator",

    fontsize=14,

    fontweight="bold"

)

plt.tight_layout()

plt.savefig(

    "LOO_MeanRankDifference.tif",

    dpi=600,

    bbox_inches="tight"

)

plt.show()

print("------------------------------------------")
print("LOO Sensitivity Analysis Completed")
print("------------------------------------------")
```

```python
# ==========================================================
# Leave-One-Indicator-Out (LOO) Robustness Matrix
# Publication Quality
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

plt.rcParams["font.family"] = "Times New Roman"

# ==========================================================
# Read Data
# ==========================================================

df = pd.read_excel("PhaseIV_Normalized_Data.xlsx")
original = pd.read_excel("PhaseVI_ARI_KBPI.xlsx")

# ==========================================================
# Indicators
# ==========================================================

indicators = [

    "Mean_Rainfall_mm",
    "Mean_Soil_Moisture",
    "Mean_NDVI",
    "Mean_NDMI",
    "Bulk_Density",
    "CEC",
    "Nitrogen",
    "SOC",
    "Irrigation_Coverage",
    "Cropping_Intensity"

]

# ==========================================================
# Entropy Weight Function
# ==========================================================

def entropy_weights(data):

    X = data.copy()

    X = X.replace(0,1e-12)

    P = X.div(X.sum(axis=0),axis=1)

    n = len(X)

    k = 1/np.log(n)

    entropy = -k*(P*np.log(P)).sum(axis=0)

    d = 1-entropy

    w = d/d.sum()

    return w

# ==========================================================
# District Order
# ==========================================================

district_order = list(

    original.sort_values(

        "KBPI_Rank"

    )["District"]

)

# ==========================================================
# Empty Matrix
# ==========================================================

heatmap = pd.DataFrame(

    index=indicators,

    columns=district_order,

    dtype=float

)

# ==========================================================
# LOO Analysis
# ==========================================================

for removed in indicators:

    print("Removing:", removed)

    remaining = [

        x

        for x in indicators

        if x != removed

    ]

    temp = df[remaining].copy()

    weights = entropy_weights(temp)

    ari = (temp * weights).sum(axis=1)

    kbpi = 1 - ari

    rank = (

        kbpi

        .rank(

            ascending=False,

            method="dense"

        )

        .astype(int)

    )

    compare = pd.DataFrame({

        "District":df["District"],

        "NewRank":rank

    })

    compare = compare.merge(

        original[["District","KBPI_Rank"]],

        on="District"

    )

    compare["RankChange"] = (

        compare["KBPI_Rank"]

        -

        compare["NewRank"]

    ).abs()

    compare = compare.set_index("District")

    heatmap.loc[removed] = compare.loc[

        district_order,

        "RankChange"

    ].values

# ==========================================================
# Save Excel
# ==========================================================

heatmap.to_excel(

    "LOO_Robustness_Matrix.xlsx"

)

print("Matrix exported.")

# ==========================================================
# Discrete Color Map
# ==========================================================

colors = [

    "#FFFFFF",   # 0

    "#C6EFCE",   # 1

    "#FFE699",   # 2

    "#F4B183",   # 3

    "#E06666",   # 4

    "#C00000"    # 5

]

cmap = ListedColormap(colors)

bounds = [-0.5,0.5,1.5,2.5,3.5,4.5,5.5]

norm = BoundaryNorm(bounds,cmap.N)

# ==========================================================
# Figure
# ==========================================================

fig, ax = plt.subplots(

    figsize=(14,6)

)

im = ax.imshow(

    heatmap.values,

    cmap=cmap,

    norm=norm,

    aspect="auto"

)

# ==========================================================
# Axis Labels
# ==========================================================

ax.set_xticks(

    np.arange(len(district_order))

)

ax.set_xticklabels(

    district_order,

    rotation=45,

    ha="right",

    fontsize=14,

    fontweight="bold"

)

ax.set_yticks(

    np.arange(len(indicators))

)

ax.set_yticklabels(

    indicators,

    fontsize=14,

    fontweight="bold"

)

# ==========================================================
# Annotate Cells
# ==========================================================

for i in range(heatmap.shape[0]):

    for j in range(heatmap.shape[1]):

        value = int(heatmap.iloc[i,j])

        ax.text(

            j,

            i,

            value,

            ha="center",

            va="center",

            fontsize=9,

            fontweight="bold"

        )

# ==========================================================
# Colorbar
# ==========================================================

cbar = plt.colorbar(

    im,

    ticks=[0,1,2,3,4,5]

)

cbar.set_label(

    "Absolute Rank Change",

    fontsize=14,

    fontweight="bold"

)

# ==========================================================
# Titles
# ==========================================================

ax.set_title(

    "Leave-One-Indicator-Out Robustness Matrix",

    fontsize=15,

    fontweight="bold"

)

ax.set_xlabel(

    "District",

    fontsize=14,

    fontweight="bold"

)

ax.set_ylabel(

    "Removed Indicator",

    fontsize=14,

    fontweight="bold"

)

plt.tight_layout()

plt.savefig(

    "LOO_Robustness_Matrix.tif",

    dpi=600,

    bbox_inches="tight"

)

plt.show()

print("Finished Successfully.")
```
